```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sentence_transformers import SentenceTransformer
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# ফাইল লোড (r স্ট্রিং নিশ্চিত করা হয়েছে)
df = pd.read_csv(r"C:\Django\AI_News_Scraper\news_data.csv")
df.dropna(subset=["Title"], inplace=True)
df.drop_duplicates(subset=["Title"], inplace=True)
df.reset_index(drop=True, inplace=True)

titles = df["Title"].tolist()
print(f"Loaded {len(titles)} headlines.")

# এম্বেডিংস তৈরি
embedder = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = embedder.encode(titles, show_progress_bar=True)

# ক্লাস্টারিং
num_clusters = 5
kmeans = KMeans(n_clusters=num_clusters, random_state=42, n_init=10)
df["Cluster_ID"] = kmeans.fit_predict(embeddings)

# PCA 2D প্রজেকশন
pca = PCA(n_components=2, random_state=42)
reduced = pca.fit_transform(embeddings)
df["PCA1"] = reduced[:, 0]
df["PCA2"] = reduced[:, 1]

# ভিজ্যুয়ালাইজেশন
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="PCA1",
    y="PCA2",
    hue="Cluster_ID",
    style="Source",
    palette="tab10",
    s=80,
)
plt.title("News Topic Clusters")
plt.show()
```

```python
target_cluster = 0  # যে ক্লাস্টারটি দেখতে চান (0, 1, 2, 3 বা 4)

cluster_data = df[df["Cluster_ID"] == target_cluster]

print(f"Total news in Cluster {target_cluster}: {len(cluster_data)}\n")

for idx, (_, row) in enumerate(cluster_data.iterrows(), start=1):
    print(f"{idx}. [{row['Source']} | {row['Category']}]")
    print(f"   Title: {row['Title']}")
    print(f"   URL:   {row['URL']}\n")
```
